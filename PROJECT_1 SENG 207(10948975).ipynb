```python
mean_alcohol=df.groupby('pH')['alcohol'].mean()
plt.bar(mean_alcohol.index, mean_alcohol)
plt.title('Mean alcohol by pH content')
plt.xlabel('pH')
plt.ylabel('alcohol(%)')
plt.show()
```

```python
mean_sulphates=df.groupby('alcohol')['sulphates'].mean()
plt.bar(mean_alcohol.index, mean_alcohol)
plt.title('Mean sulphates by alcohol content')
plt.xlabel('alcohol')
plt.ylabel('sulphates(%)')
plt.show()
```

```python
mean_alcohol=df.groupby('quality')['alcohol'].mean()
plt.bar(mean_alcohol.index, mean_alcohol)
plt.title('Mean alcohol by quality content')
plt.xlabel('quality')
plt.ylabel('alcohol(%)')
plt.show()
```
